# Lesson 2.1 - Terraform Providers, Resources, and State

**Level: Beginner**

## Learning Objectives
- Understand provider, resource, and state concepts
- Create a first Terraform configuration
- Run `init` and `plan` safely

Reference: [Terraform State](https://developer.hashicorp.com/terraform/language/state)

```python
%%writefile work/terraform/main.tf
terraform {
  required_version = ">= 1.6.0"
}

resource "local_file" "hello" {
  filename = "hello-iac.txt"
  content  = "Hello from Terraform"
}
```

## Code-Along
Use local resources first so you can learn workflow without cloud credentials.

```python
%%bash
cd work/terraform
echo "Run in your terminal when Terraform is installed:"
echo "terraform init"
echo "terraform plan"
```

## What This Produces
- `work/terraform/main.tf`
- A planned local file resource and Terraform state metadata
