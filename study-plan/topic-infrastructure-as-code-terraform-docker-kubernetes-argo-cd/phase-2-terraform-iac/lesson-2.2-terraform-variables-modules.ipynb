**Level: Intermediate**

## Learning Objectives
- Parameterize Terraform code with variables
- Create a basic reusable module
- Understand outputs for composition

Reference: [Terraform Modules](https://developer.hashicorp.com/terraform/language/modules)

```python
%%bash
mkdir -p work/terraform/modules/text_file
```

```python
%%writefile work/terraform/modules/text_file/main.tf
variable "filename" { type = string }
variable "content"  { type = string }

resource "local_file" "this" {
  filename = var.filename
  content  = var.content
}

output "written_file" {
  value = local_file.this.filename
}
```

```python
%%writefile work/terraform/module-usage.tf
module "notes" {
  source   = "./modules/text_file"
  filename = "module-note.txt"
  content  = "Generated from a Terraform module"
}
```
