## Suggested Workflow
1. Terraform plans infra changes in PR
2. Merge triggers apply in controlled environment
3. App manifest changes are synced by Argo CD
4. Promote by changing image tag or overlay in Git

Reference: [Argo CD Declarative Setup](https://argo-cd.readthedocs.io/en/stable/operator-manual/declarative-setup/)

```python
%%writefile work/repo/WORKFLOW.md
# Lightweight IaC + GitOps Workflow
- PR: terraform fmt + validate + plan
- Merge: approved apply for infra
- PR: kubernetes manifest update
- Argo CD auto-sync: cluster convergence
- Promotion: dev -> staging by Git change
```
