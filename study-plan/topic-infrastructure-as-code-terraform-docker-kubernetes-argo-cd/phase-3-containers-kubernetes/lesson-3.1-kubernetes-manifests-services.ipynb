# Lesson 3.1 - Kubernetes Manifests and Services

**Level: Intermediate**

## Learning Objectives
- Define Deployment, Service, and ConfigMap manifests
- Apply updates through declarative YAML
- Verify workload rollout status

Reference: [Kubernetes Services](https://kubernetes.io/docs/concepts/services-networking/service/)

```python
%%writefile work/k8s/app-stack.yaml
apiVersion: v1
kind: ConfigMap
metadata:
  name: app-config
data:
  APP_MESSAGE: "hello from configmap"
---
apiVersion: apps/v1
kind: Deployment
metadata:
  name: app
spec:
  replicas: 2
  selector:
    matchLabels:
      app: app
  template:
    metadata:
      labels:
        app: app
    spec:
      containers:
      - name: app
        image: nginx:1.27-alpine
        ports:
        - containerPort: 80
---
apiVersion: v1
kind: Service
metadata:
  name: app-svc
spec:
  selector:
    app: app
  ports:
  - port: 80
    targetPort: 80
  type: ClusterIP
```

```python
%%bash
echo "Run in your terminal:"
echo "kubectl apply -f work/k8s/app-stack.yaml"
echo "kubectl rollout status deploy/app"
echo "kubectl get pods,svc"
```
