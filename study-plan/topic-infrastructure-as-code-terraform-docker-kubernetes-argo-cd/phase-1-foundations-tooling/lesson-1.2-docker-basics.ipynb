## Concepts
A **Docker image** is a template. A **container** is a running instance of that template.

```python
%%writefile work/docker/Dockerfile
FROM nginx:1.27-alpine
COPY index.html /usr/share/nginx/html/index.html
EXPOSE 80
```

```python
%%writefile work/docker/index.html
<html><body><h1>IaC Learning Track</h1><p>Docker lesson output.</p></body></html>
```

```python
%%bash
cd work/docker
echo "Run in your terminal if Docker is available:"
echo "docker build -t iac-demo:1 ."
echo "docker run --rm -p 8080:80 iac-demo:1"
```

## What This Produces
- `work/docker/Dockerfile`
- `work/docker/index.html`
- A runnable local web container (when Docker is available)
