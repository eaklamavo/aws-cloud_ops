## Concepts
**Infrastructure as Code (IaC)** means you describe infrastructure in files, then tools create it consistently.

Think of IaC like a cooking recipe: same ingredients and steps produce the same dish every time.

```python
%%bash
mkdir -p work/{terraform,docker,k8s,argocd}
echo "Created workspace folders:"
find work -maxdepth 2 -type d | sort
```

## Code-Along: Create a tool check script
This script helps you verify if your core tools are available before moving into labs.

```python
%%writefile work/tool-check.sh
#!/usr/bin/env bash
set -e
for cmd in terraform docker kubectl git; do
  if command -v "$cmd" >/dev/null 2>&1; then
    echo "[OK] $cmd found"
  else
    echo "[WARN] $cmd missing"
  fi
done
```

```python
%%bash
chmod +x work/tool-check.sh
./work/tool-check.sh
```
