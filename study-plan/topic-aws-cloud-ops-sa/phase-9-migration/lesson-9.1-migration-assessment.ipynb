## The 7Rs Migration Strategies

| Strategy | What it means | Effort | Value |
|----------|--------------|--------|-------|
| **Retire** | Decommission — no longer needed | None | $$$ |
| **Retain** | Keep on-premises for now (revisit later) | None | $ |
| **Rehost** | Lift-and-shift to EC2 | Low | $$ |
| **Replatform** | Lift-tinker-shift (minor optimizations) | Medium | $$$ |
| **Repurchase** | Move to SaaS | Medium | $$$$ |
| **Refactor/Re-architect** | Rebuild cloud-native | High | $$$$$ |
| **Relocate** | Move to AWS using VMware Cloud on AWS | Low-Med | $$ |

## Migration Portfolio Assessment
Before migrating, inventory your workloads:
1. **Discover**: AWS Application Discovery Service scans on-prem servers
2. **Assess**: Score each app by complexity, criticality, migration strategy
3. **Prioritize**: Wave planning — migrate easy apps first for quick wins
4. **Plan**: Wave 1 (low complexity) → Wave 2 → Wave 3 (complex/critical)

## TCO (Total Cost of Ownership) Analysis
```
On-premises cost includes:
- Hardware purchase + refresh
- Data center (power, cooling, rent)
- IT staff (admins, networking)
- Software licenses
- DR infrastructure

AWS cost:
- Compute (EC2 / containers / Lambda)
- Storage (EBS / S3 / EFS)
- Networking (data transfer)
- Support plan
```
AWS Pricing Calculator: https://calculator.aws

````python
# Simulate a migration portfolio assessment
import json

# Example application inventory for assessment
applications = [
    {"name": "E-commerce Frontend", "type": "Web App", "env": "prod",
     "complexity": "low", "criticality": "high", "db": "RDS MySQL",
     "recommended_strategy": "Replatform", "reason": "Containerize for scalability"},
    {"name": "HR Management System", "type": "Legacy App", "env": "prod",
     "complexity": "high", "criticality": "medium", "db": "Oracle",
     "recommended_strategy": "Retain then Refactor", "reason": "Oracle complexity needs careful planning"},
    {"name": "Dev/Test Servers", "type": "Infrastructure", "env": "dev",
     "complexity": "low", "criticality": "low", "db": "None",
     "recommended_strategy": "Rehost", "reason": "Lift-and-shift for quick win, rightsizing later"},
    {"name": "Email Server", "type": "Collaboration", "env": "prod",
     "complexity": "medium", "criticality": "medium", "db": "None",
     "recommended_strategy": "Repurchase", "reason": "Move to Amazon WorkMail or SaaS"},
    {"name": "Old Reporting Tool", "type": "Analytics", "env": "prod",
     "complexity": "low", "criticality": "low", "db": "MSSQL",
     "recommended_strategy": "Retire", "reason": "Replaced by new BI tool"},
]

print("Migration Portfolio Assessment:")
print(f"{'Application':30} {'Strategy':25} {'Complexity':12} {'Reason'}")
print("-" * 100)
for app in applications:
    print(f"{app['name']:30} {app['recommended_strategy']:25} {app['complexity']:12} {app['reason']}")

# Calculate migration waves
waves = {"Wave 1 (Quick wins)": [], "Wave 2 (Standard)": [], "Wave 3 (Complex)": []}
for app in applications:
    if app['recommended_strategy'] in ['Retire', 'Repurchase', 'Rehost'] and app['complexity'] == 'low':
        waves["Wave 1 (Quick wins)"].append(app['name'])
    elif app['complexity'] == 'high':
        waves["Wave 3 (Complex)"].append(app['name'])
    else:
        waves["Wave 2 (Standard)"].append(app['name'])

print("\nMigration Wave Plan:")
for wave, apps in waves.items():
    print(f"  {wave}: {', '.join(apps) or 'None'}") 
````

## Summary
- **7Rs**: Retire → Retain → Rehost → Replatform → Repurchase → Refactor → Relocate
- Start with **Retire** (removes cost immediately) and **Rehost** (quick wins, low risk)
- **Wave planning**: Easy apps first → build experience → tackle complex apps in later waves
- **TCO analysis**: Include hidden on-prem costs (hardware refresh, staff, DR, licenses)
